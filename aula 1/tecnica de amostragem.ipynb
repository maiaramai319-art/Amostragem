A amostragem é o pilar que sustenta a validade científica de uma pesquisa experimental. No estudo de cimentos ecoeficientes (LC3) submetidos a diferentes temperaturas de cura, a estratégia de amostragem estratificada permite dividir o experimento em grupos controlados (ex: 15%, 30% e 45% de argila), garantindo que a variabilidade dos materiais não mascare os efeitos reais das adições minerais (OLIVEIRA, 2020).

Durante uma fase laboratorial, uma amostragem pesquisada simples é obrigatória para a seleção dos corpos de prova que serão rompidos em idades específicas (7, 28 e 91 dias), obedecendo às diretrizes da NBR 7215 (ABNT, 2019) para eliminação do inquérito do pesquisador. Complementarmente, uma amostragem sistemática é fundamental em análises de microestrutura (DRX/MEV), permitindo a coleta de dados em intervalos fixos de profundidade para mapear o gradiente de hidratação (SCRIVENER et al., 2018).
